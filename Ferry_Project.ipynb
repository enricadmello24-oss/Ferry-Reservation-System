```python
import csv

MAINT_FILE = "maintenance_records.csv"
TICKET_FILE = "ticket_records.csv"


# ---------------- Helpers ----------------

def get_next_id(filename):
    try:
        with open(filename, "r") as f:
            rows = list(csv.reader(f))

            if rows:
                return str(int(rows[-1][0]) + 1)
            else:
                return "1"

    except FileNotFoundError:
        return "1"


# ---------------- Report Printing ----------------

def print_maintenance_report(row):
    print("=" * 40)
    print(f"      MAINTENANCE REPORT - {row[0]}")
    print(f"  Ferry ID               : {row[1]}")
    print(f"  Date                   : {row[2]}")
    print(f"  Status                 : {row[3]}")
    print(f"  Cost                   : {row[4]}")
    print("=" * 40)
    print()


def print_ticket_report(row):
    print("=" * 40)
    print(f"        TICKET REPORT   - {row[0]}")
    print(f"  User Name             : {row[1]}")
    print(f"  Ferry Name            : {row[2]}")
    print(f"  Seat No               : {row[3]}")
    print(f"  Fare                  : {row[4]}")
    print(f"  Status                : {row[5]}")
    print(f"  Payment               : {row[6]}")
    print("=" * 40)
    print()


# ---------------- Maintenance ----------------

def add_maintenance():
    mid = input("Enter Maintenance ID: ")

    # Check for duplicate ID
    try:
        with open(MAINT_FILE, "r") as f:
            for row in csv.reader(f):
                if row and row[0] == mid:
                    raise ValueError(
                        f"Maintenance ID '{mid}' already exists! "
                        "Enter a unique ID."
                    )
    except FileNotFoundError:
        pass  # Add new record

    fid = input("Enter Ferry ID: ")
    date = input("Enter Date (DD-MM-YYYY): ")
    status = input("Enter Status (planned/in progress/completed): ")
    cost = input("Enter Cost: ")

    with open(MAINT_FILE, "a", newline="") as f:
        csv.writer(f).writerow([mid, fid, date, status, cost])

    print(f"Maintenance record added with ID: {mid}\n")


def display_maintenance():
    try:
        with open(MAINT_FILE, "r") as f:
            rows = list(csv.reader(f))

            if not rows:
                print("No maintenance records found.\n")
                return

            for row in rows:
                print_maintenance_report(row)

    except FileNotFoundError:
        print("No maintenance records found.\n")


def view_one_maintenance():
    mid = input("Enter Maintenance ID: ")

    try:
        with open(MAINT_FILE, "r") as f:
            found = False

            for row in csv.reader(f):
                if row and row[0] == mid:
                    found = True
                    print_maintenance_report(row)
                    break

            if not found:
                print("Record not found.\n")

    except FileNotFoundError:
        print("No maintenance records found.\n")


def update_maintenance():
    mid = input("Enter Maintenance ID to update: ")

    rows = []
    updated = False

    try:
        with open(MAINT_FILE, "r") as f:
            for row in csv.reader(f):
                if row and row[0] == mid:
                    print("Old Record:")
                    print_maintenance_report(row)

                    print("1. Update Date")
                    print("2. Update Status")
                    print("3. Update Cost")

                    choice = input("Enter choice: ")

                    if choice == "1":
                        row[2] = input("New Date (DD-MM-YYYY): ")
                    elif choice == "2":
                        row[3] = input(
                            "New Status (planned/in progress/completed): "
                        )
                    elif choice == "3":
                        row[4] = input("New Cost: ")

                    updated = True

                rows.append(row)

        if updated:
            with open(MAINT_FILE, "w", newline="") as f:
                csv.writer(f).writerows(rows)

            print("Record updated successfully.\n")
        else:
            print("Record not found.\n")

    except FileNotFoundError:
        print("No maintenance records found.\n")


def delete_maintenance():
    mid = input("Enter Maintenance ID to delete: ")

    rows = []
    deleted = False

    try:
        with open(MAINT_FILE, "r") as f:
            for row in csv.reader(f):
                if row and row[0] == mid:
                    deleted = True
                else:
                    rows.append(row)

        if deleted:
            with open(MAINT_FILE, "w", newline="") as f:
                csv.writer(f).writerows(rows)

            print("Record deleted successfully.\n")
        else:
            print("Record not found.\n")

    except FileNotFoundError:
        print("No maintenance records found.\n")


# ---------------- Tickets ----------------

def book_ticket():
    tid = get_next_id(TICKET_FILE)

    uname = input("Enter User Name: ")
    fname = input("Enter Ferry Name: ")
    seat = input("Enter Seat No: ")
    fare = input("Enter Fare: ")
    status = input("Enter Status (confirmed/waiting): ")
    pay = input("Enter Payment (paid/unpaid): ")

    with open(TICKET_FILE, "a", newline="") as f:
        csv.writer(f).writerow([tid, uname, fname, seat, fare, status, pay])

    print(f"Ticket booked with ID: {tid}\n")


def delete_ticket():
    tid = input("Enter Ticket ID to delete: ")

    rows = []
    deleted = False

    try:
        with open(TICKET_FILE, "r") as f:
            for row in csv.reader(f):
                if row and row[0] == tid:
                    deleted = True
                else:
                    rows.append(row)

        if deleted:
            with open(TICKET_FILE, "w", newline="") as f:
                csv.writer(f).writerows(rows)

            print("Ticket deleted successfully.\n")
        else:
            print("Ticket not found.\n")

    except FileNotFoundError:
        print("No tickets found.\n")


def display_tickets():
    try:
        with open(TICKET_FILE, "r") as f:
            rows = list(csv.reader(f))

            if not rows:
                print("No tickets found.\n")
                return

            for row in rows:
                print_ticket_report(row)

    except FileNotFoundError:
        print("No tickets found.\n")


# ---------------- Main Menu ----------------

while True:
    print("\n~ Ferry Reservation System ~")
    print("1. Add Maintenance Record")
    print("2. Display All Maintenance Records")
    print("3. View One Maintenance Record")
    print("4. Update Maintenance Record")
    print("5. Delete Maintenance Record")
    print("6. Book Ticket")
    print("7. Delete Ticket")
    print("8. Display All Tickets")
    print("9. Exit")

    choice = input("Enter choice (1-9): ")

    if choice == "1":
        add_maintenance()
    elif choice == "2":
        display_maintenance()
    elif choice == "3":
        view_one_maintenance()
    elif choice == "4":
        update_maintenance()
    elif choice == "5":
        delete_maintenance()
    elif choice == "6":
        book_ticket()
    elif choice == "7":
        delete_ticket()
    elif choice == "8":
        display_tickets()
    elif choice == "9":
        print("Exiting... Thank you for using the Ferry Reservation System!")
        break
    else:
        print("Invalid choice! Try again.\n")
```
